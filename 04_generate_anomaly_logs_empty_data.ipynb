```python
import pandas as pd
from pathlib import Path
columns = [
    "detection_time",
    "anomaly_type",
    "ip",
    "path",
    "metric_value",
    "threshold",
    "severity",
    "date",
    "detail",
]

df = pd.DataFrame(columns=columns)

output_dir = Path("output")
output_dir.mkdir(exist_ok=True)
output_path = output_dir / "04_anomaly_logs.parquet"
df.to_parquet(output_path, index=False)
print(f"已生成空 Parquet：{output_path}")
```
